```python
def scrape_reviews(url, xpath):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    tree = html.fromstring(response.content)
    reviews = tree.xpath(xpath)
    
    return [review.strip() for review in reviews if isinstance(review, str)]
```

```python
import requests
from bs4 import BeautifulSoup

def scrape_full_text(url):
    try:
        # Send request
        headers = {
            "User-Agent": "Mozilla/5.0"
        }
        response = requests.get(url, headers=headers)
        response.raise_for_status()

        # Parse HTML
        soup = BeautifulSoup(response.text, "html.parser")

        # Remove script & style elements
        for script in soup(["script", "style", "noscript"]):
            script.extract()

        # Get visible text
        text = soup.get_text(separator="\n")

        # Clean text (remove extra spaces)
        lines = [line.strip() for line in text.splitlines()]
        clean_text = "\n".join(line for line in lines if line)

        return clean_text

    except Exception as e:
        return f"Error: {e}"



url = "https://www.denofgeek.com/movies/avatar-fire-and-ash-review-james-cameron-shallow/"


full_text_denofgeek = scrape_full_text(url)

# Print output
print(full_text_denofgeek)
```

```python
def extract_reviews_and_save(all_texts, output_file="Text Files/avatar_review.txt"):
    
    combined_text = "\n".join(all_texts)

    lines = combined_text.split("\n")
    reviews = []

    unwanted_keywords = [
        "login", "settings", "log out", "home", "subscribe",
        "menu", "search", "sign in", "notifications", "Abhimanyu Mathur"
    ]

    for line in lines:
        line = line.strip()

        if not line:
            continue

        # Skip unwanted UI text
        if any(word in line.lower() for word in unwanted_keywords):
            continue


        if len(line.split()) > 25:
            reviews.append(line)


    reviews = list(set(reviews))

    # Save to file
    with open(output_file, "w", encoding="utf-8") as f:
        for i, review in enumerate(reviews, 1):
            f.write(f"{i}. {review}\n\n")

    print(f"Saved {len(reviews)} reviews to {output_file}")


all_sources = [
    full_text_denofgeek,
    full_text_bollymoviereviewz,
]


try:
    all_sources.append("\n".join(wogma_reviews))
except:
    pass

try:
    all_sources.append("\n".join(commonsense_reviews))
except:
    pass


extract_reviews_and_save(all_sources)
```
