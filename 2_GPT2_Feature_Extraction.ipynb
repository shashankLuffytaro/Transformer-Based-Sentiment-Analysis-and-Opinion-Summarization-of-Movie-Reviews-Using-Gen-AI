```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
pd.set_option("display.max_columns", None)
import matplotlib.pyplot as plt
%matplotlib inline
from tqdm import tqdm
from lib_file import lib_path
import re
import torch
from transformers import GPT2Model, GPT2Tokenizer
```

***Load Cleaned Data***

```python
df = pd.read_csv("input/cleaned_data.csv")
df.head()
```

```python
df['sentiment'].value_counts()
```

```python
# Load a pre-trained GPT-2 model
gpt2_model = GPT2Model.from_pretrained('gpt2')
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
```

```python
texts = df['cleaned_text'].values.tolist()
texts[:5]
```

```python
%%time

features = []
for text in tqdm(texts):
    inputs = tokenizer(text, return_tensors="pt")
    with torch.no_grad():
        output = gpt2_model(**inputs)
        features.append(output.last_hidden_state.mean(dim=1).squeeze().tolist())
```

```python
new_df = pd.DataFrame(features, columns=[f"column_{i+1}" for i in range(len(features[0]))])
new_df['target'] = df['sentiment'].values
new_df.head()
```

```python
new_df.to_csv("extracted_features.csv", index=False)
```
